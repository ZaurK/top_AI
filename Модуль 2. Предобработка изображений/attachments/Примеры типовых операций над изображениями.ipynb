```python
"""
В этой ячейке выполняется начальная загрузка файлов изображений
для работы примеров кода из репозитория в заданную папку
(в коде задана папка Colab "/content").
Достаточно ячейку запустить на выполнение один раз.

Можно также просто перетащить уже скачанные файлы изображений
в папку Colab "/content". Тогда можно данную ячейку не запускать.

"""

import os
import requests
from urllib.parse import unquote

def download_files(file_urls, output_dir='downloaded'):
    """
    Загружает несколько файлов из списка URL и сохраняет их в указанную папку.

    Args:
        file_urls (list): список raw-ссылок на файлы в GitHub.
        output_dir (str): имя папки для сохранения (создаётся, если не существует).
    """
    # Создаём папку, если её нет
    os.makedirs(output_dir, exist_ok=True)

    for url in file_urls:
        try:
            # Отправляем GET-запрос
            response = requests.get(url, stream=True, timeout=10)
            response.raise_for_status()  # выбрасывает исключение при статусе 4xx/5xx

            # Извлекаем имя файла из URL (последняя часть после /)
            filename = unquote(url.split('/')[-1])  # декодируем %20 и т.п.
            local_path = os.path.join(output_dir, filename)

            # Сохраняем содержимое в файл
            with open(local_path, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    if chunk:
                        f.write(chunk)

            print(f'✅ Загружен: {filename} -> {local_path}')

        except requests.exceptions.RequestException as e:
            print(f'❌ Ошибка при загрузке {url}: {e}')

# Пример использования:
if __name__ == '__main__':
    # Список raw-ссылок на файлы (замените на свои)
    urls = [
        'https://github.com/ZaurK/top_AI/blob/main/%D0%9C%D0%BE%D0%B4%D1%83%D0%BB%D1%8C%202.%20%D0%9F%D1%80%D0%B5%D0%B4%D0%BE%D0%B1%D1%80%D0%B0%D0%B1%D0%BE%D1%82%D0%BA%D0%B0%20%D0%B8%D0%B7%D0%BE%D0%B1%D1%80%D0%B0%D0%B6%D0%B5%D0%BD%D0%B8%D0%B9/attachments/Dachi_01.jpg',
        'https://github.com/ZaurK/top_AI/blob/main/%D0%9C%D0%BE%D0%B4%D1%83%D0%BB%D1%8C%202.%20%D0%9F%D1%80%D0%B5%D0%B4%D0%BE%D0%B1%D1%80%D0%B0%D0%B1%D0%BE%D1%82%D0%BA%D0%B0%20%D0%B8%D0%B7%D0%BE%D0%B1%D1%80%D0%B0%D0%B6%D0%B5%D0%BD%D0%B8%D0%B9/attachments/%D0%9B%D0%B0%D0%BD%D0%B4%D1%88%D0%B0%D1%84%D1%82%20ANN%20-%2001.png',
        'https://github.com/ZaurK/top_AI/blob/main/%D0%9C%D0%BE%D0%B4%D1%83%D0%BB%D1%8C%202.%20%D0%9F%D1%80%D0%B5%D0%B4%D0%BE%D0%B1%D1%80%D0%B0%D0%B1%D0%BE%D1%82%D0%BA%D0%B0%20%D0%B8%D0%B7%D0%BE%D0%B1%D1%80%D0%B0%D0%B6%D0%B5%D0%BD%D0%B8%D0%B9/attachments/%D0%9B%D0%B0%D0%BD%D0%B4%D1%88%D0%B0%D1%84%D1%82%20ANN%20-%2002.png',
        'https://github.com/ZaurK/top_AI/blob/main/%D0%9C%D0%BE%D0%B4%D1%83%D0%BB%D1%8C%202.%20%D0%9F%D1%80%D0%B5%D0%B4%D0%BE%D0%B1%D1%80%D0%B0%D0%B1%D0%BE%D1%82%D0%BA%D0%B0%20%D0%B8%D0%B7%D0%BE%D0%B1%D1%80%D0%B0%D0%B6%D0%B5%D0%BD%D0%B8%D0%B9/attachments/%D0%9D%D0%B5%D0%B4%D0%B2%D0%B8%D0%B6%D0%B8%D0%BC%D0%BE%D1%81%D1%82%D1%8C_01.jpg',
    ]

    download_files(urls, output_dir='/content')
```

```python
"""
Гистограмма яркости
"""
import numpy as np
import matplotlib.pyplot as plt

from skimage.util import img_as_ubyte
from skimage import data
from skimage.exposure import histogram

noisy_image = img_as_ubyte(data.camera())
hist, hist_centers = histogram(noisy_image)

fig, ax = plt.subplots(ncols=2, figsize=(10, 5))

ax[0].imshow(noisy_image, cmap=plt.cm.gray)
ax[0].set_axis_off()

ax[1].plot(hist_centers, hist, lw=2)
ax[1].set_title('Gray-level histogram')

fig.tight_layout()
```

```python
"""
Бинаризация изображений

"""


import matplotlib
import matplotlib.pyplot as plt

from skimage.data import page
from skimage.filters import threshold_otsu, threshold_niblack, threshold_sauvola
from skimage.exposure import histogram

matplotlib.rcParams['font.size'] = 9


image = page()
binary_global = image > threshold_otsu(image)


hist, hist_centers = histogram(image)
plt.title('Гистограмма яркости')
plt.plot(hist_centers, hist, lw=2)


window_size = 25
thresh_niblack = threshold_niblack(image, window_size=window_size, k=0.8)
thresh_sauvola = threshold_sauvola(image, window_size=window_size)

binary_niblack = image > thresh_niblack
binary_sauvola = image > thresh_sauvola

plt.figure(figsize=(8, 7))
plt.subplot(3, 2, 1)
plt.imshow(image, cmap=plt.cm.gray)
plt.title('Оригинал')
plt.axis('off')

plt.subplot(3, 2, 2)
plt.title('Глобальный порог')
plt.imshow(binary_global, cmap=plt.cm.gray)
plt.axis('off')

plt.subplot(3, 2, 3)
plt.imshow(binary_niblack, cmap=plt.cm.gray)
plt.title('Метод Ниблэка')
plt.axis('off')

plt.subplot(3, 2, 4)
plt.imshow(binary_sauvola, cmap=plt.cm.gray)
plt.title('Метод Саувола')
plt.axis('off')

plt.subplot(3, 2, 5)
plt.title('Порог Ниблэка')
plt.imshow(thresh_niblack, cmap=plt.cm.gray)
plt.axis('off')

plt.subplot(3, 2, 6)
plt.title('Порог Саувола')
plt.imshow(thresh_sauvola, cmap=plt.cm.gray)
plt.axis('off')

plt.show()
```

```python
"""
Подадвление импульсного шума
"""

from skimage.filters.rank import median
from skimage.morphology import disk, ball
from PIL import Image

rng = np.random.default_rng()
noise = rng.random(noisy_image.shape)
noisy_image = img_as_ubyte(data.camera())
print(type(noisy_image), noisy_image.shape)
noisy_image[noise > 0.99] = 255
noisy_image[noise < 0.01] = 0

fig, axes = plt.subplots(2, 2, figsize=(10, 10), sharex=True, sharey=True)
ax = axes.ravel()

ax[0].imshow(noisy_image, vmin=0, vmax=255, cmap=plt.cm.gray)
ax[0].set_title('Noisy image')

ax[1].imshow(median(noisy_image, disk(1)), vmin=0, vmax=255, cmap=plt.cm.gray)
ax[1].set_title('Median $r=1$')

ax[2].imshow(median(noisy_image, disk(5)), vmin=0, vmax=255, cmap=plt.cm.gray)
ax[2].set_title('Median $r=5$')

ax[3].imshow(median(noisy_image, disk(20)), vmin=0, vmax=255, cmap=plt.cm.gray)
ax[3].set_title('Median $r=20$')

for a in ax:
    a.set_axis_off()

fig.tight_layout()
```

```python
"""
Афинные преобразования по трем точкам

"""

img = cv2.imread('Dachi_01.jpg')
print(img.shape)
show_image(img)

src_tri = np.array( [[75, 75], [200, 320], [650, 200]] ).astype(np.float32)
dst_tri = np.array( [[0, 0], [0, 423], [1000, 211]] ).astype(np.float32)

warp_mat = cv2.getAffineTransform(src_tri, dst_tri)
dst_image = cv2.warpAffine(img, warp_mat, (1200, 798))
show_image(dst_image)

```

```python
"""
Перспективные преобразования
"""

img = cv2.imread("Dachi_01.jpg")
print(img.shape)
show_image(img)

# Задаем координаты четверки точек для исходного и результирующего изображения
pts1 = np.float32([[100,100],[1000,100],[100,600],[1000,600]])
pts2 = np.float32([[200,200],[900,200],[100,600],[1000,600]])

# Вычисляем матрицу гомографии
M = cv2.getPerspectiveTransform(pts1,pts2)

# Деформируем изображение
dst = cv2.warpPerspective(img,M,(1200,798))
show_image(dst)
```
